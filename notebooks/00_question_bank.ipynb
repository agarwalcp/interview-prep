## Known limitation (say this in your viva)
- **Backend** answers come with real paragraph text → keywords are extracted from a real answer.
- **Other roles**: their source repos give only a question title or an external link, not an answer, so keywords come from the *question text itself*. A candidate repeating the question's words can score — keyword grading is weakest there. Mitigation: hand-edit the `expected_keywords` column of `data/question_bank.csv` for the questions you demo.
